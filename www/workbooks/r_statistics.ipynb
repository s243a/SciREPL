# Statistics with R

Hypothesis testing, regression, and confidence intervals using built-in R datasets.

No data download or package installation needed — uses base R only.

## 1. Descriptive Statistics

```r
data(mtcars)
cat("Dataset: mtcars (", nrow(mtcars), "cars, ", ncol(mtcars), "variables)\n\n")
summary(mtcars[, c("mpg", "hp", "wt", "disp")])
```

## 2. Two-Sample t-Test

Do manual-transmission cars get better MPG than automatics?

```r
auto <- mtcars$mpg[mtcars$am == 0]
manual <- mtcars$mpg[mtcars$am == 1]

cat("Automatic:", round(mean(auto), 1), "mpg (n =", length(auto), ")\n")
cat("Manual:   ", round(mean(manual), 1), "mpg (n =", length(manual), ")\n\n")

t_result <- t.test(manual, auto, alternative = "greater")
print(t_result)

cat("\nConclusion:",
    ifelse(t_result$p.value < 0.05,
           "Reject H0 — manual cars have significantly higher MPG",
           "Fail to reject H0"))
```

## 3. Chi-Squared Test

Are cylinder count and transmission type independent?

```r
tab <- table(Cylinders = mtcars$cyl, Transmission = mtcars$am)
colnames(tab) <- c("Automatic", "Manual")
print(tab)
cat("\n")
chisq.test(tab)
```

## 4. Multiple Linear Regression

```r
model <- lm(mpg ~ wt + hp + am, data = mtcars)
summary(model)
```

## 5. Regression Diagnostics

```r
par(mfrow = c(2, 2))
plot(model)
```

## 6. Confidence Intervals

```r
ci <- confint(model, level = 0.95)
cat("95% Confidence Intervals:\n")
print(round(ci, 4))
```

```r
coefs <- coef(model)[-1]
ci_vals <- ci[-1, ]
n <- length(coefs)

par(mfrow = c(1, 1), mar = c(5, 8, 4, 2))
plot(coefs, 1:n, xlim = range(ci_vals),
     pch = 19, col = "#58a6ff", cex = 1.5,
     yaxt = "n", xlab = "Estimate", ylab = "",
     main = "95% Confidence Intervals for Coefficients")
axis(2, at = 1:n, labels = names(coefs), las = 1)
segments(ci_vals[, 1], 1:n, ci_vals[, 2], 1:n,
         lwd = 3, col = "#58a6ff")
abline(v = 0, lty = 2, col = "#f85149", lwd = 1.5)
```

## 7. One-Way ANOVA

Does MPG differ significantly across cylinder counts?

```r
anova_model <- aov(mpg ~ factor(cyl), data = mtcars)
summary(anova_model)
cat("\nTukey HSD Post-hoc Comparisons:\n")
TukeyHSD(anova_model)
```

```r
boxplot(mpg ~ cyl, data = mtcars,
        main = "MPG by Cylinder Count",
        xlab = "Cylinders", ylab = "Miles per Gallon",
        col = c("#58a6ff", "#a371f7", "#f85149"))
```

## Summary

- **Welch t-test**: In the unadjusted one-sided comparison, manual-transmission cars have higher mean MPG
- **Chi-squared**: The contingency table suggests an association, but the small expected counts trigger an approximation warning, so interpret this result cautiously
- **Regression**: Weight and horsepower are significant negative predictors after adjustment; transmission type is not significant in this model
- **ANOVA**: MPG differs significantly among the 4-, 6-, and 8-cylinder groups; the Tukey results identify the pairwise differences
